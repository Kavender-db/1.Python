```python
# print 0 to 20 by using range

for num in range (0,20):
    print(num)
```

```python
# print range 10 to 20

for num in range (10,20):
    print(num, end=" ")
print()
```

```python
# Print number of items in the list by using 'len'

find_total_num = [10, 20, 14, 55, 43, 87, 76]
print(find_total_num)

print("Number of item in the List2:", len(find_total_num))
```
