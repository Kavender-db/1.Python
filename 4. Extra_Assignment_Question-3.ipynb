```python
# catagory the people by their age like children, adult, citizen, senior citizen

age= int(input("age:"))

if (1 <= age <= 17):
    print("Child")
elif (18 <= age <= 30):
    print("adult")
elif (31 <= age <= 45):
    print("citizen")
elif (46 <= age):
    print("senior citizen")
else:
    print("invalid input")
```
